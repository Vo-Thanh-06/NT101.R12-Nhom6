```python

# NHIEM VU 2.1 - CAU 3
# Ma hoa thong diep bang RSA - Truong hop 2

import base64

p = 20079993872842322116151219
q = 676717145751736242170789
e = 17

n = p * q
phi = (p - 1) * (q - 1)
d = pow(e, -1, phi)

message = "The University of Information Technology"
message_bytes = message.encode("utf-8")

# Xac dinh kich thuoc moi khoi
block_size = (n.bit_length() - 1) // 8
cipher_size = (n.bit_length() + 7) // 8

# Chia thong diep thanh cac khoi
blocks = [
    message_bytes[i:i + block_size]
    for i in range(0, len(message_bytes), block_size)
]

# Ma hoa RSA tung khoi
encrypted = b""
for block in blocks:
    m = int.from_bytes(block, "big")
    c = pow(m, e, n)
    encrypted += c.to_bytes(cipher_size, "big")

# Luu do dai thong diep de phuc vu giai ma
payload = len(message_bytes).to_bytes(4, "big") + encrypted

# Chuyen ban ma sang Base64
cipher_base64 = base64.b64encode(payload).decode("ascii")

# Giai ma kiem tra
decoded = base64.b64decode(cipher_base64)
total_length = int.from_bytes(decoded[:4], "big")
cipher_data = decoded[4:]

decrypted = b""
for i in range(0, len(cipher_data), cipher_size):
    cipher_block = cipher_data[i:i + cipher_size]
    c = int.from_bytes(cipher_block, "big")
    m = pow(c, d, n)
    size = min(block_size, total_length - len(decrypted))
    decrypted += m.to_bytes(size, "big")

decrypted_message = decrypted.decode("utf-8")

print("===== CAU 3 - RSA TRUONG HOP 2 =====")
print("Thong diep goc:", message)
print("So byte thong diep:", len(message_bytes))
print("Kich thuoc khoi du lieu:", block_size)
print("So khoi du lieu:", len(blocks))
print("Ban ma Base64:", cipher_base64)
print("Thong diep sau giai ma:", decrypted_message)
print("Kiem tra:", message == decrypted_message)
```

```python

# NHIEM VU 2.1 - CAU 4
# Giai ma ban ma thu nhat bang RSA

import base64

# Cap khoa RSA thu hai
p = 20079993872842322116151219
q = 676717145751736242170789
e = 17

n = p * q
phi = (p - 1) * (q - 1)
d = pow(e, -1, phi)

# Ban ma Base64 tu de bai
cipher_text = (
    "raUcesUlOkx/8ZhgodMoo0Uu18sC20yXlQFevSu7W/"
    "FDxIy0YRHMyXcHdD9PBvIT2aUft5fCQEGomiVVPv4I"
)

# Chuyen Base64 thanh byte
cipher_bytes = base64.b64decode(cipher_text)

# Chia du lieu thanh cac khoi RSA
block_size = (n.bit_length() + 7) // 8
blocks = [
    cipher_bytes[i:i + block_size]
    for i in range(0, len(cipher_bytes), block_size)
]

# Giai ma tung khoi voi thu tu byte little-endian
plaintext_bytes = b""

for block in blocks:
    c = int.from_bytes(block, "little")
    m = pow(c, d, n)
    decoded_block = m.to_bytes(block_size - 1, "little")
    plaintext_bytes += decoded_block.rstrip(b"\x00")

# Chuyen thanh chuoi van ban
plaintext = plaintext_bytes.decode("utf-8")

print("===== CAU 4 - GIAI MA BAN MA THU NHAT =====")
print("Cap khoa su dung: RSA truong hop 2")
print("Kich thuoc moi khoi:", block_size, "byte")
print("So khoi ban ma:", len(blocks))
print("Ban ro tim duoc:", plaintext)
```

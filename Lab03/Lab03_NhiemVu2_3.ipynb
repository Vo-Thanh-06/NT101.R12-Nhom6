```python
%%bash
set -e

apt-get update -qq
apt-get install -y -qq git build-essential

if [ ! -d /content/md5collgen ]; then
    git clone https://github.com/zhijieshi/md5collgen.git /content/md5collgen
fi

make -C /content/md5collgen

if [ -x /content/md5collgen/md5collgen ]; then
    echo "CAI DAT MD5COLLGEN THANH CONG!"
else
    echo "CAI DAT THAT BAI!"
fi
```

```python
%%bash
mkdir -p /content/Lab03_NV23
cd /content/Lab03_NV23

# Tao tien to 9 byte
printf 'UIT-Lab03' > prefix.txt

echo "=== KICH THUOC TIEN TO ==="
wc -c prefix.txt

# Tao hai tep va cham MD5
/content/md5collgen/md5collgen -p prefix.txt -o out1.bin out2.bin

echo "=== SO SANH HAI TEP ==="
diff out1.bin out2.bin

echo "=== GIA TRI BAM MD5 ==="
md5sum out1.bin out2.bin

echo "=== KICH THUOC HAI TEP ==="
wc -c out1.bin out2.bin
```
